# Pitch

Expand topic into short pitch, identifying:

* The misconception it corrects
* The payoff quantity
* Why it needs motion

````python
# Define file paths
topic_file_path = os.path.join("..", "docs", "topic.txt")
pitch_file_path = os.path.join(os.path.dirname(topic_file_path), "pitch.txt")

# Read the topic from the topic.txt file
try:
    with open(topic_file_path, "r") as file:
        topic = file.read().strip()
    print(f"Loaded Topic: {topic}\n")
    print("-" * 40)
except FileNotFoundError:
    print(f"Error: Could not find {topic_file_path}. Please ensure the file exists.")
    topic = None

if topic:
    # Define the prompt template
    prompt_template = PromptTemplate(
        input_variables=["topic"],
        template="""
        You are an expert educational animation producer. 
        Take the following single-sentence technical idea and expand it into a short pitch for a 60 to 90-second animation.
        
        Topic: {topic}
        
        Your pitch must explicitly identify and output the following three elements:
        1. Misconception: The specific misconception this animation corrects.
        2. Payoff Quantity: A measurable payoff (a number, count, size, or comparison that can be computed and asserted).
        3. Why It Needs Motion: What specifically moves in the animation (since a static definition gains nothing from animation).
        """
    )

    # Format the prompt and invoke the agent
    prompt = prompt_template.format(topic=topic)
    response = llm.invoke(prompt)
    
    pitch_content = response.content
    
    # FIX: Safely parse the response into a string if LangChain returns a list of blocks
    if isinstance(pitch_content, list):
        pitch_content = "".join([
            block.get("text", "") if isinstance(block, dict) else str(block) 
            for block in pitch_content
        ])
        
    print(pitch_content)
    print("-" * 40)
    
    # Save the generated pitch back to the docs directory
    try:
        with open(pitch_file_path, "w") as file:
            file.write(pitch_content)
        print(f"Success: Pitch saved to {pitch_file_path}")
    except IOError as e:
        print(f"Error writing to file: {e}")
````

# Reference Scene

````python
if os.path.exists(pitch_file_path):
    with open(pitch_file_path, "r", encoding="utf-8", errors="replace") as file:
        pitch_content = file.read()
    print(f"Loaded pitch from {pitch_file_path}")
else:
    raise FileNotFoundError(f"Could not find pitch_content in memory or in {pitch_file_path}. Please run the 'Pitch' cell first.")

# 2. Instruct Gemini to generate the 6-scene storyboard and the executable Manim code
storyboard_prompt = PromptTemplate(
    input_variables=["pitch"],
    template="""
    You are an expert Manim animator and director. Read the following animation pitch:
    {pitch}

    First, break this pitch down into exactly a 6-scene storyboard. For each scene, provide:
    - The Visual Concept
    - The Duration in seconds (the total across all 6 scenes must equal 60 to 90 seconds).

    Second, write the complete, functioning Python code using the Manim library to render this exact 6-scene storyboard.
    - The main class MUST be named `ExponentialSnowball`.
    - Rely on standard Manim shapes (Squares, Circles, Rectangles, Axes, Text, VGroup) and math functions to represent the concepts safely.
    - CRITICAL MANIM CE SYNTAX: NEVER use the `alignment` keyword argument in Text or any Mobject (it is deprecated and will crash). If you need to align text, use `.next_to()`, `.shift()`, `.move_to()`, or `text_alignment=...` instead.
    - ALWAYS wrap hex color codes in string quotes (e.g., color="#2ecc71" or COLOR = "#2ecc71").
    - Ensure all animations use `self.play()` and `self.wait()` to match your specified scene durations.
    - Output ONLY the Python code in a single markdown block (```python ... ```) at the very end of your response.
    """
)

print("Requesting 6-scene storyboard and Manim generation from Gemini...")
prompt_text = storyboard_prompt.format(pitch=pitch_content)
response = llm.invoke(prompt_text)
output = response.content

# Safely parse the response into a string if LangChain returns a list of blocks
if isinstance(output, list):
    output = "".join([
        block.get("text", "") if isinstance(block, dict) else str(block) 
        for block in output
    ])

# 3. Separate the storyboard text from the generated Python code
storyboard_text = output.split("```python")[0]
print("\n--- Storyboard Breakdown ---")
print(storyboard_text.strip())
print("----------------------------\n")

# 4. Extract the Python code from the markdown block
code_match = re.search(r"```python\n(.*?)```", output, re.DOTALL)
if code_match:
    manim_code = code_match.group(1).strip()
    
    # Safety Net 1: Wrap unquoted hex colors in quotes
    manim_code = re.sub(r'=\s*(#[0-9a-fA-F]{6})', r'= "\1"', manim_code)
    # Safety Net 2: Forcefully strip out any 'alignment=...' arguments to prevent TypeErrors
    manim_code = re.sub(r',\s*alignment\s*=\s*[a-zA-Z0-9_\.\'"]+', '', manim_code)
    
    # 5. Write the generated code to a local script
    script_filename = "generated_manim_scene.py"
    with open(script_filename, "w", encoding="utf-8") as f:
        if "from manim import *" not in manim_code:
            f.write("from manim import *\n\n")
        f.write(manim_code)
        
    print(f"Manim code extracted and saved to {script_filename}. Rendering MP4 (this will take a few minutes)...")
    
    # 6. Execute Manim using the current kernel's Python executable (.venv)
    render_result = subprocess.run(
        [sys.executable, "-m", "manim", "-qm", "-v", "WARNING", script_filename, "ExponentialSnowball"],
        capture_output=True,
        text=True,
        encoding="utf-8",
        errors="replace" 
    )
    
    if render_result.returncode == 0:
        print("Render completed successfully!")
        
        video_path = os.path.join("media", "videos", "generated_manim_scene", "720p30", "ExponentialSnowball.mp4")
        
        if os.path.exists(video_path):
            # 7. Display and autoplay the generated MP4 inside the Jupyter Notebook cell
            display(Video(video_path, embed=True, html_attributes="controls autoplay loop style='max-width: 100%;'"))
        else:
            print(f"Error: Render succeeded but expected video file was not found at {video_path}")
    else:
        print("Manim encountered an error during rendering:\n", render_result.stderr)
else:
    print("Error: Gemini did not return a valid Python code block.")
````
