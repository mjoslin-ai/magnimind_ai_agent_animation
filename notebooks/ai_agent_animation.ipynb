# Test Model Call

````python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")


def get_text(response) -> str:
    """Extract string content from LangChain LLM responses safely."""
    content = response.content
    if isinstance(content, list):
        return "".join(
            block.get("text", "") if isinstance(block, dict) else str(block)
            for block in content
        )
    return str(content)


# Test connection
print(get_text(llm.invoke("Say 'Model connection successful.'")))
````

# Create Directories

````python
DOCS_DIR = Path("../docs")
SRC_DIR = Path("../src")

for folder in (DOCS_DIR, SRC_DIR):
    folder.mkdir(parents=True, exist_ok=True)
````

# Topic

````python
topic_file = DOCS_DIR / "topic.txt"

# Specify what to add to the text file
topic_content = """Visualize the geometric growth of an S&P 500 ETF at an average ten percent compound growth rate to demonstrate how a single investment can multiply many times over the decades, revealing why starting early creates far more wealth than trying to time the market."""

topic_file.write_text(topic_content.strip(), encoding="utf-8")
````

# Pitch

Expand topic into short pitch, identifying:

* The misconception it corrects
* The payoff quantity
* Why it needs motion

````python
pitch_file = DOCS_DIR / "pitch.txt"

# Generate pitch
pitch_prompt = f"""
You are an expert educational animation producer. 
Take the following single-sentence technical idea and expand it into a short pitch for a 60 to 90-second animation.

Topic: {topic_file.read_text(encoding='utf-8')}

Your pitch must explicitly identify and output the following four elements:
1. Title: A descriptive PascalCase name to be used for the Manim class and file name (e.g., SnpGrowthMultiplier).
2. Misconception: The specific misconception this animation corrects.
3. Payoff Quantity: A measurable payoff (a number, count, size, or comparison that can be computed and asserted).
4. Why It Needs Motion: What specifically moves in the animation.
"""

pitch_content = get_text(llm.invoke(pitch_prompt))
pitch_file.write_text(pitch_content, encoding="utf-8")
print(f"Success: Pitch saved to {pitch_file}")
````

# Reference Scene

````python
# Extract the Title generated in the pitch
class_match = re.search(r"(?:Title|title)[^\w]*([A-Z][a-zA-Z0-9_]+)", pitch_content)
main_class = class_match.group(1) if class_match else "GeneratedScene"

storyboard_prompt = f"""
You are an expert Manim animator and director. Read the following animation pitch:
{pitch_content}

First, break this pitch down into exactly a 6-scene storyboard. For each scene, provide:
- The Visual Concept
- The Duration in seconds (the total across all 6 scenes must equal 60 to 90 seconds).

Second, write the complete, functioning Python code using the Manim library to render this exact 6-scene storyboard.
- The main class MUST be named `{main_class}`.
- Rely on standard Manim shapes (Squares, Circles, Rectangles, Axes, Text, VGroup) and math functions to represent the concepts safely.
- CRITICAL MANIM CE SYNTAX: NEVER use the `alignment` keyword argument in Text or any Mobject (it is deprecated and will crash). If you need to align text, use `.next_to()`, `.shift()`, `.move_to()`, or `text_alignment=...` instead.
- ALWAYS wrap hex color codes in string quotes (e.g., color="#2ecc71" or COLOR = "#2ecc71").
- Ensure all animations use `self.play()` and `self.wait()` to match your specified scene durations.
- Output ONLY the Python code in a single markdown block (```python ... ```) at the very end of your response.
"""

print("Requesting 6-scene storyboard and Manim generation from Gemini...")
output = get_text(llm.bind(temperature=0.3).invoke(storyboard_prompt))

# Print storyboard
storyboard_text = output.split("```python")[0].strip()
print(
    f"\n--- Storyboard Breakdown ---\n{storyboard_text}\n----------------------------\n"
)

# Extract and clean code block
if code_match := re.search(r"```python\n(.*?)```", output, re.DOTALL):
    manim_code = code_match.group(1).strip()
    manim_code = re.sub(r"=\s*(#[0-9a-fA-F]{6})", r'= "\1"', manim_code)
    manim_code = re.sub(
        r",\s*alignment\s*=\s*[a-zA-Z0-9_\.\'\"]+", "", manim_code
    )

    if "from manim import *" not in manim_code:
        manim_code = "from manim import *\n\n" + manim_code

    script_path = Path("generated_manim_scene.py")
    script_path.write_text(manim_code, encoding="utf-8")
    print(
        f"Manim code saved to {script_path}. Rendering {main_class} to MP4 video..."
    )

    # Render video via subprocess
    cmd = [
        sys.executable,
        "-m",
        "manim",
        "-qm",
        "-v",
        "WARNING",
        "--media_dir",
        "..",
        str(script_path),
        main_class,
    ]
    render_result = subprocess.run(
        cmd, capture_output=True, text=True, encoding="utf-8", errors="replace"
    )

    video_path = Path(
        f"../videos/generated_manim_scene/720p30/{main_class}.mp4"
    )
    if render_result.returncode == 0 and video_path.exists():
        print("Render completed successfully!")
        display(
            Video(
                str(video_path),
                embed=True,
                html_attributes="controls autoplay loop style='max-width: 100%;'",
            )
        )
    else:
        print("Manim error during rendering:\n", render_result.stderr)
else:
    print("Error: Gemini did not return a valid Python code block.")
````
